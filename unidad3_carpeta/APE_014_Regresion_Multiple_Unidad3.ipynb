# **📊 Tarea 1: Ajuste del Modelo de Regresión Lineal Múltiple**



## **Objetivo**

En esta práctica se implementa un modelo de **Regresión Lineal Múltiple** utilizando el método de **Mínimos Cuadrados Ordinarios (OLS)** con la biblioteca **Statsmodels**. El propósito es predecir el **Tiempo de Compilación de un Software (Y)** a partir de dos variables predictoras:

- **Líneas de Código (X₁):** representa la cantidad de líneas desarrolladas en el software.
- **Memoria RAM (X₂):** cantidad de memoria disponible durante el proceso de compilación.
- **Tiempo de Compilación (Y):** tiempo necesario para completar la compilación.

El procedimiento consiste en crear un conjunto de datos simulado, definir las variables independientes y dependiente, agregar el término constante (intercepto), entrenar el modelo de regresión múltiple y presentar el resumen estadístico mediante `summary()`. Finalmente, se generan predicciones y una comparación gráfica entre los valores observados y los estimados para evaluar el comportamiento del modelo.

```python
# ==========================================================
# TAREA 1: AJUSTE DEL MODELO DE REGRESIÓN MÚLTIPLE SIMULADO
# ==========================================================

# Conjunto de datos simulado
datos = pd.DataFrame({
    "Lineas_Codigo": [1200,1500,1700,2000,2200,2500,2700,3000,3300,3600],
    "RAM": [4,4,8,8,16,16,32,32,32,64],
    "Tiempo_Compilacion": [14,16,18,21,24,27,31,35,38,42]
})

# Visualizar el conjunto de datos
display(datos)

# Variables predictoras (X) y variable respuesta (Y)
X = datos[["Lineas_Codigo", "RAM"]]
Y = datos["Tiempo_Compilacion"]

# Agregar la constante
X = sm.add_constant(X)

# Ajustar el modelo de regresión múltiple
modelo = sm.OLS(Y, X).fit()

# Resumen estadístico
print(modelo.summary())

# Predicciones
datos["Tiempo_Predicho"] = modelo.predict(X)

# Visualización: Valores reales vs. valores predichos
plt.figure(figsize=(8,5))

plt.plot(
    datos.index,
    datos["Tiempo_Compilacion"],
    marker="o",
    linewidth=2,
    label="Valores Reales"
)

plt.plot(
    datos.index,
    datos["Tiempo_Predicho"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Valores Predichos"
)

plt.title("Comparación entre Valores Reales y Predichos")
plt.xlabel("Observaciones")
plt.ylabel("Tiempo de Compilación")
plt.legend()
plt.grid(True)

plt.show()
```

# **🔬 Tarea 2: Hito del Proyecto - Modelado Predictivo Regional (ABP)**

## **Objetivo**

En esta actividad se desarrolla un modelo de **Regresión Lineal Múltiple** utilizando el dataset regional del Proyecto Integrador sobre casos de dengue. Previamente se realiza la limpieza de los datos eliminando registros con valores nulos en las variables seleccionadas.

La variable respuesta corresponde a la **Edad** de los pacientes, mientras que las variables predictoras son la **Semana Epidemiológica (SE)** y el **Año** de registro. Posteriormente se ajusta un modelo mediante el método de **Mínimos Cuadrados Ordinarios (OLS)** y se analiza el resumen estadístico para evaluar la influencia de cada predictor sobre la variable respuesta.

```python
# ==========================================================
# TAREA 2: MODELADO PREDICTIVO REGIONAL
# ==========================================================

# Cargar el dataset
df = pd.read_excel(
    "Datos_Dengue_MSP_Ene2021_Ago2025(1).xlsx",
    sheet_name="DCSA_DG_2021_2025"
)

# Seleccionar las variables de estudio y eliminar valores nulos
datos = df[["Edad", "SE", "Año"]].dropna()

# Visualizar los primeros registros
display(datos.head())

# Definir variables predictoras (X) y variable respuesta (Y)
X = datos[["SE", "Año"]]
Y = datos["Edad"]

# Agregar la constante (intercepto)
X = sm.add_constant(X)

# Ajustar el modelo de regresión múltiple
modelo = sm.OLS(Y, X).fit()

# Mostrar el resumen estadístico
print(modelo.summary())

# Generar predicciones
datos["Edad_Predicha"] = modelo.predict(X)

# Mostrar una comparación de valores reales y predichos
display(datos[["Edad", "Edad_Predicha"]].head())

plt.figure(figsize=(12,6))

plt.plot(
    datos.index,
    datos["Edad"],
    color="#2563EB",
    linewidth=3,
    marker="o",
    markersize=7,
    label="Edad Real"
)

plt.plot(
    datos.index,
    datos["Edad_Predicha"],
    color="#EF4444",
    linewidth=3,
    marker="^",
    markersize=7,
    linestyle="--",
    label="Edad Predicha"
)

plt.fill_between(
    datos.index,
    datos["Edad"],
    datos["Edad_Predicha"],
    color="#93C5FD",
    alpha=0.25
)

plt.title(
    "Comparación entre la Edad Observada y la Edad Estimada",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel("Número de Observación", fontsize=12)
plt.ylabel("Edad (años)", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.grid(True, alpha=0.25)

plt.legend(
    loc="upper left",
    fontsize=11,
    frameon=True,
    shadow=True
)

plt.tight_layout()

plt.show()
```

# **📌 Tarea 4: ABI - Diagnóstico de Multicolinealidad (VIF)**




## **Objetivo**

En esta actividad se realiza un diagnóstico avanzado de **multicolinealidad** mediante el **Factor de Inflación de la Varianza (Variance Inflation Factor - VIF)**, uno de los indicadores más utilizados en el análisis de modelos de regresión lineal múltiple para evaluar la dependencia existente entre las variables predictoras.

La multicolinealidad es un fenómeno estadístico que ocurre cuando dos o más variables independientes presentan una elevada correlación entre sí, compartiendo parte de la misma información. Esta situación dificulta que el modelo identifique el efecto individual de cada predictor sobre la variable respuesta, ya que resulta complejo distinguir cuál de las variables está explicando realmente la variación observada.

Aunque la multicolinealidad no reduce necesariamente la capacidad predictiva del modelo, sí afecta la calidad de las estimaciones de los coeficientes de regresión. Entre sus principales consecuencias se encuentran el incremento de la varianza de los coeficientes estimados, el aumento de los errores estándar, la ampliación de los intervalos de confianza, la disminución de la significancia estadística de algunas variables y una mayor sensibilidad del modelo frente a pequeñas modificaciones en los datos. Como resultado, los coeficientes pueden volverse inestables e incluso cambiar de magnitud o de signo, dificultando su interpretación.

Para detectar este problema se utiliza el **Factor de Inflación de la Varianza (VIF)**, el cual cuantifica cuánto aumenta la varianza de un coeficiente de regresión debido a la correlación existente con las demás variables independientes. Matemáticamente, el VIF de una variable se obtiene mediante la siguiente expresión:

$$[
VIF_i=\frac{1}{1-R_i^2}
]$$

donde:

- $(VIF_i)$ representa el Factor de Inflación de la Varianza de la variable $(i)$.
- $(R_i^2)$ corresponde al coeficiente de determinación obtenido al regredir la variable $(X_i)$ contra el resto de variables predictoras.

De acuerdo con esta expresión, cuando una variable presenta una alta dependencia con las demás, el valor de $(R_i^2)$ se aproxima a **1**, provocando un incremento considerable del VIF y evidenciando la presencia de multicolinealidad. Por el contrario, valores de VIF cercanos a **1** indican que la variable aporta información propia al modelo y que no existe una dependencia lineal significativa con los demás predictores.

### **Criterios de Interpretación del Factor de Inflación de la Varianza (VIF)**

| Rango del VIF | Nivel de Riesgo | Interpretación | Acción Recomendada |
|:-------------:|:---------------:|----------------|--------------------|
| **VIF = 1** | Nulo | La variable es completamente independiente del resto de predictores y no presenta correlación lineal. | No es necesario realizar ninguna acción. |
| **1 < VIF < 5** | Bajo | Existe una correlación débil o moderada entre las variables predictoras, sin afectar significativamente la estabilidad del modelo. | La variable puede mantenerse en el modelo. |
| **5 ≤ VIF < 10** | Moderado | Se observa una multicolinealidad que puede comenzar a influir en la precisión de los coeficientes estimados. | Analizar la conveniencia de eliminar o transformar variables. |
| **VIF ≥ 10** | Alto | Existe una multicolinealidad severa que compromete la estabilidad e interpretación del modelo de regresión. | Se recomienda replantear el modelo eliminando, combinando o transformando variables predictoras. |

```python
# ==========================================================
# TAREA 4: DIAGNÓSTICO DE MULTICOLINEALIDAD (VIF)
# ==========================================================

from statsmodels.stats.outliers_influence import variance_inflation_factor

predictoras = ["SE", "Año"]

X_vif = datos[predictoras].dropna()
X_vif = sm.add_constant(X_vif)

# ----------------------------------------------------------
# Cálculo del VIF
# ----------------------------------------------------------
vif = pd.DataFrame({
    "Variable": X_vif.columns,
    "VIF": [variance_inflation_factor(X_vif.values, i)
            for i in range(X_vif.shape[1])]
})
vif["VIF"] = vif["VIF"].round(3)

def interpretar_vif(valor):
    if valor < 1.5:
        return "Muy baja"
    elif valor < 5:
        return "Baja"
    elif valor < 10:
        return "Moderada"
    else:
        return "Alta (crítica)"

vif["Diagnóstico"] = vif["VIF"].apply(interpretar_vif)

# Tabla completa (incluye const, solo por transparencia)
print("Factor de Inflación de la Varianza (VIF)\n")
display(vif)

# Vista solo de predictores reales, ordenada por severidad
vif_predictores = (
    vif[vif["Variable"] != "const"]
    .sort_values("VIF", ascending=False)
    .reset_index(drop=True)
)

# ----------------------------------------------------------
# Visualización
# ----------------------------------------------------------
colores = {
    "Muy baja": "#4CAF50",
    "Baja": "#8BC34A",
    "Moderada": "#FFC107",
    "Alta (crítica)": "#F44336"
}
color_barras = vif_predictores["Diagnóstico"].map(colores)

plt.figure(figsize=(7, 4))

barras = plt.bar(
    vif_predictores["Variable"],
    vif_predictores["VIF"],
    color=color_barras,
    edgecolor="black",
    linewidth=1.2
)

plt.axhline(y=5, color="darkorange", linestyle="--", linewidth=1.5,
            label="VIF = 5 (moderado)")
plt.axhline(y=10, color="red", linestyle="--", linewidth=1.5,
            label="VIF = 10 (crítico)")

for barra in barras:
    altura = barra.get_height()
    plt.text(barra.get_x() + barra.get_width() / 2, altura + 0.15,
              f"{altura:.2f}", ha="center", fontsize=10, weight="bold")

plt.title("Factor de Inflación de la Varianza (VIF)", fontsize=14, weight="bold")
plt.xlabel("Variables Predictoras")
plt.ylabel("Valor del VIF")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------------------------------------
# 5. Conclusión
# ----------------------------------------------------------
criticas = vif_predictores[vif_predictores["VIF"] >= 10]["Variable"].tolist()
moderadas = vif_predictores[
    (vif_predictores["VIF"] >= 5) & (vif_predictores["VIF"] < 10)
]["Variable"].tolist()

print("\n--- Conclusión del diagnóstico ---\n")
if criticas:
    print(f"Las variables {criticas} presentan multicolinealidad crítica "
          f"(VIF ≥ 10). Se recomienda eliminarlas o combinarlas mediante "
          f"una técnica de reducción de dimensionalidad (ej. PCA).")
elif moderadas:
    print(f"Las variables {moderadas} presentan multicolinealidad moderada "
          f"(5 ≤ VIF < 10). No es obligatorio eliminarlas, pero conviene "
          f"vigilar la estabilidad de sus coeficientes.")
else:
    print("No se detecta multicolinealidad relevante entre los predictores "
          "(todos los VIF < 5). El modelo puede interpretarse con confianza.")
```

## **Interpretación del Diagnóstico de Multicolinealidad**

El Factor de Inflación de la Varianza $(VIF)$ cuantifica el grado en que la varianza de un coeficiente de regresión se ve inflada como consecuencia de la correlación lineal que dicha variable mantiene con el resto de los predictores del modelo. Formalmente, se define como $VIF = 1 / (1 − R²ᵢ)$, donde $R²ᵢ$ corresponde al coeficiente de determinación obtenido al regresar la variable *i* sobre las demás predictoras. Valores elevados de $VIF$ señalan que una variable comparte información redundante con otras, comprometiendo la estabilidad e interpretabilidad de los coeficientes estimados por Mínimos Cuadrados Ordinarios.

En este estudio se calcularon los valores de $VIF$ para las variables **SE (Semana Epidemiológica)** y **Año**, obteniéndose resultados inferiores al umbral crítico de referencia $(VIF < 5)$ establecido en la literatura estadística [1]. Este hallazgo constituye evidencia de que ambas variables aportan información estadísticamente independiente al modelo, sin indicios de colinealidad estructural entre ellas.

Este resultado es coherente con el análisis exploratorio desarrollado en la Tarea 3, donde la matriz de correlación reportó un coeficiente de **r = −0.18** entre **SE** y **Año**, valor que —según los criterios convencionales de interpretación $(|r| < 0.3 = correlación débil)$ descarta una relación lineal fuerte entre ambos predictores. La convergencia entre el análisis bivariado $(correlación de Pearson)$ y el diagnóstico multivariado $(VIF)$ refuerza la validez y consistencia interna del diagnóstico.

Desde una perspectiva estadística, la ausencia de multicolinealidad relevante implica que:

- Cada predictor contribuye con información marginal propia al poder explicativo del modelo.
- Los errores estándar de los coeficientes $β$ no se encuentran artificialmente inflados, lo que preserva la validez de las pruebas de hipótesis individuales $(columna P>|t|)$.
- La interpretación *ceteris paribus* de cada coeficiente el efecto de una variable manteniendo constante la otra resulta estadísticamente defendible dentro de los límites del modelo.

Con base en el análisis conjunto de correlación bivariada $(Tarea 3)$ y el diagnóstico de multicolinealidad mediante $VIF$ $(Tarea 4)$, se concluye que el modelo de regresión múltiple presenta una estructura estable y libre de redundancia significativa entre sus predictores. Las variables **SE** y **Año** pueden mantenerse simultáneamente en la especificación del modelo, dado que no existe evidencia estadística de dependencia lineal relevante entre ellas.

## **5. Preguntas de Control**

### **1. Considerando la naturaleza de la causalidad, ¿cómo la lógica de 'mantener las demás variables constantes' permite aislar el efecto de un factor específico en un fenómeno complejo, y qué limitaciones filosóficas impone esto a la realidad?**

En una regresión múltiple, el coeficiente $βᵢ$ se interpreta bajo la cláusula *ceteris paribus*: el efecto de $xᵢ$ sobre Y cuando el resto de predictores permanece fijo [1]. Esta lógica es matemáticamente equivalente a tomar una derivada parcial en lugar de una derivada total, lo que permite aislar estadísticamente la contribución marginal de un factor dentro de un sistema donde, en la realidad, varias causas actúan de forma simultánea [2].

La limitación filosófica es que esta separación es **analítica, no ontológica**: en el mundo real las variables rara vez varían de forma verdaderamente independiente, por lo que la cláusula ceteris paribus expresa una tendencia bajo condiciones idealizadas y no una ley universal e incondicional [1]. El modelo no elimina la interdependencia del fenómeno; simplemente la neutraliza matemáticamente para fines de estimación, lo cual es útil para inferencia estadística pero requiere cautela al momento de extrapolar relaciones causales fuertes a partir de datos puramente observacionales [2].

### **2. Si el conocimiento del modelo crece al añadir más datos, ¿por qué añadir variables irrelevantes ('basura') engaña al ajuste del modelo? ¿Qué nos enseña esta paradoja sobre la diferencia entre la complejidad aparente y la verdad subyacente?**

El $R²$ ordinario nunca disminuye al añadir predictores, incluso si estos son ruido estadístico, porque siempre existe algo de covarianza muestral aleatoria entre cualquier variable nueva y el error [3]. El $R²$ Ajustado corrige este sesgo introduciendo una penalización basada en los grados de libertad $(n − k − 1):$ si la variable añadida no reduce la suma de cuadrados residual más de lo esperado por azar, el $R²$ Ajustado se mantiene igual o disminuye [3].

Esta paradoja enseña que la **complejidad aparente no equivale a conocimiento real**: un modelo puede "verse" más completo por tener más variables, mientras en realidad solo está absorbiendo ruido de la muestra específica, degradando su capacidad de generalizar a datos nuevos (sobreajuste) [3]. La verdad subyacente del fenómeno no crece con el número de variables; crece únicamente cuando cada variable añadida aporta información genuinamente nueva y no redundante.

### **3. Ante la prueba F-statistic global, ¿es posible que un modelo sea estadísticamente significativo pero carezca de propósito práctico? ¿Cómo se reconcilia la precisión matemática con la incertidumbre del mundo real en la toma de decisiones de ingeniería?**

Sí es posible: un modelo puede rechazar la hipótesis nula del $F-statistic global$ (es decir, ser estadísticamente significativo) y aun así tener escaso valor práctico. Esto ocurre principalmente cuando el tamaño de muestra es grande, ya que incluso relaciones triviales en magnitud producen p-valores muy bajos [4]. La significancia estadística responde a la pregunta "¿existe el efecto?", mientras que la significancia práctica evaluada mediante el tamaño del efecto y no solo el p-valor responde a "¿ese efecto importa lo suficiente para la toma de decisiones?" [4].

La reconciliación en ingeniería consiste en no detenerse en el resultado de la prueba de hipótesis: se debe complementar con la magnitud del coeficiente, su intervalo de confianza y el contexto de aplicación (costos, tolerancia del sistema, umbrales operativos), de modo que la precisión matemática del modelo se traduzca en un criterio de decisión útil, no solo en una afirmación formalmente correcta [4].

### **4. Si dos causas parecen moverse siempre juntas (multicolinealidad), ¿podemos realmente distinguir sus efectos por separado? ¿Qué nos dice este problema geométrico/matricial sobre las limitaciones de nuestro entendimiento al intentar separar fenómenos interdependientes?**

Cuando dos predictores están altamente correlacionados, geométricamente sus vectores en el espacio de las variables independientes son casi colineales; la matriz $(XᵀX)$ se vuelve casi singular y su inversa de la cual dependen los coeficientes estimados se torna numéricamente inestable [5]. El $VIF$ cuantifica exactamente esto: mide cuánto se infla la varianza de un coeficiente porque su predictor puede explicarse en gran medida por los demás $(VIFⱼ = 1/(1−R²ⱼ))$ [5].

En términos prácticos, cuando la multicolinealidad es severa, el modelo **no puede distinguir confiablemente** cuál de las dos variables correlacionadas es la responsable del efecto sobre Y; el reparto del "crédito explicativo" entre ambas se vuelve arbitrario y sensible a pequeñas variaciones en los datos [5]. Este problema geométrico/matricial ilustra un límite epistemológico más amplio: cuando dos fenómenos co-varían casi perfectamente en los datos disponibles, ninguna herramienta puramente estadística sin información experimental o estructural adicional puede separar sus efectos individuales de forma no ambigua.

### **5. Ante la necesidad de simplificar un sistema complejo para comprenderlo, ¿es preferible sacrificar variables (información) o aceptar la imprecisión del modelo? ¿Qué define, bajo su criterio, a un 'buen' modelo para la ingeniería?**

Esta tensión es la esencia del *bias-variance tradeoff*: modelos muy simples tienden a tener mayor sesgo (subajuste) pero menor varianza, mientras que modelos muy complejos reducen el sesgo pero aumentan la varianza y el riesgo de sobreajuste [6]. El principio de parsimonia (Navaja de Occam aplicada a la modelación estadística) sugiere preferir, entre modelos con desempeño comparable, el que use menos variables o supuestos, porque generaliza mejor a datos nuevos y es más interpretable [7].

Bajo este criterio, un "buen" modelo de ingeniería no es el que maximiza el ajuste a los datos de entrenamiento, sino el que logra el mejor equilibrio entre capacidad explicativa, estabilidad de sus coeficientes y capacidad de generalización — sacrificar una variable irrelevante casi siempre es preferible a mantenerla, salvo que exista evidencia teórica sólida de que aporta información causal relevante [6], [7].

## **6. Conclusiones**

El desarrollo de esta práctica evidenció que un modelo de regresión múltiple es tan confiable como los supuestos que lo sostienen. La lógica *ceteris paribus* permite aislar el efecto individual de cada predictor, pero esta separación es un artificio analítico y no una descripción completa de la causalidad del fenómeno regional estudiado [1], [2]. Asimismo, se comprobó que el R² Ajustado actúa como un mecanismo de control frente a la inclusión de variables irrelevantes, recordando que un modelo más complejo no es necesariamente un modelo más verdadero [3].

El diagnóstico de multicolinealidad mediante VIF confirmó que las variables **SE** y **Año** no presentan redundancia estructural significativa (VIF < 5), lo que valida la estabilidad de los coeficientes estimados y respalda su interpretación individual con un nivel adecuado de confianza estadística [5]. Finalmente, se reconoce que la significancia estadística (F-statistic, valores-p) es una condición necesaria pero no suficiente para la utilidad práctica de un modelo: la decisión de mantener, simplificar o descartar un modelo en un contexto de ingeniería debe integrar tanto el rigor matemático como el juicio contextual sobre la magnitud y relevancia real de los efectos estimados [4], [6], [7].
